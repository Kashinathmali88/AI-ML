```python
# Real data

years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]

plt.plot(years,revenue)
plt.title("Movies revenue in each year")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")
```

```python
non_oscar_revenue = [95,235,45,156,541]

plt.plot(years,revenue , color = "red" ,marker = "<" , linestyle = "-.",linewidth = "3",label = "Oscare movies")
plt.plot(years,non_oscar_revenue ,"<-y", label = "Non-Oscare movies")

plt.title("Oscar movies VS Non-Oscar moveis revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")

plt.legend()
```

```python
years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]
non_oscar_revenue = [95,235,45,156,541]


plt.plot(years,revenue ,label = "Oscare movies")
plt.plot(years,non_oscar_revenue ,label = "Non-Oscare movies")

plt.title("Oscar movies VS Non-Oscar moveis revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")

print(plt.style.available)

# plt.style.use("dark_background")
plt.savefig("demo_fig.png")
```

```python
# Bar chat
years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]

plt.bar(years,revenue)
plt.title("Years VS Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue")
plt.grid()
```

```python
# Bar chat
years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]

plt.bar(years,revenue)
plt.title("Years VS Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue")

for i in range(len(years)):
    plt.text(years[i] , revenue[i]+5, revenue[i],ha="center")

plt.ylim(0,max(revenue) + 50)
```

```python
# compare to data
import numpy as np

years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]
non_oscar_revenue = [95,235,45,156,541]

plt.title("Oscar Vs Non-Oscar Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $(M)")

width =  0.4
x = np.arange(len(years))

plt.bar(x - width / 2,non_oscar_revenue,width,label="Oscar")
plt.bar(x + width / 2,non_oscar_revenue,width,label="Non-Oscar")

plt.legend()
plt.xticks(x,years)
```

```python
Age = [22, 25, 28, 30, 32, 35, 38, 40, 42, 45, 48, 50, 53, 55, 58, 60, 63, 65, 68, 70]

BG = [85, 90, 95, 100, 105, 110, 115, 120, 125, 130, 135, 140, 145, 150, 155, 160, 165, 170, 175, 180]

colors = ["green" if x < 120 else "red" for x in BG]
plt.scatter(Age,BG,color=colors,alpha=0.5)
plt.grid()

for i in range(len(Age)):
    plt.annotate(Age[i],xy=(Age[i],BG[i]),xytext=(Age[i],BG[i]))
```

```python
# commpate multiple data set

cities = [
    "Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata",
    "Hyderabad", "Pune", "Jaipur", "Ahmedabad", "Lucknow"
]

winter_temp = [24, 12, 20, 25, 18, 19, 17, 14, 16, 13]
winter_humidity = [60, 55, 65, 70, 60, 50, 55, 45, 50, 60]

summer_temp = [33, 40, 31, 37, 36, 38, 35, 41, 42, 39]
summer_humidity = [75, 35, 50, 65, 70, 40, 45, 25, 30, 40]

plt.title("Winter VS Summer Data")
plt.scatter(winter_temp,winter_humidity,label="Winter")
plt.scatter(summer_temp,summer_humidity,label="Summer")
plt.ylabel("Temprecture")
plt.xlabel("Humidity")
plt.legend()
```

```python
# Pie cahrt

expenses = [
    "Salaries",
    "Rent",
    "Marketing",
    "Utilities",
    "Software",
    "Travel",
    "Training"
]

amounts = [
    500000,
    120000,
    80000,
    30000,
    50000,
    40000,
    20000
]

explode = [0,0.3,0,0,0,0,0]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
       wedgeprops =  {
           "edgecolor":"black",
           "linewidth":1,
           "linestyle":"--",
       },
       explode=explode,
       startangle=45,
        shadow=True
       )
```
