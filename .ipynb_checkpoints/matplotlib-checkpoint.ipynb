```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [5,6,7,8]

plt.plot(x,y) # line plot
plt.show()
```

```python
# Real data

years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]

plt.plot(years,revenue)
plt.title("Movies revenue in each year")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")
```

```python
non_oscar_revenue = [95,235,45,156,541]

plt.plot(years,revenue , color = "red" ,marker = "<" , linestyle = "-.",linewidth = "3",label = "Oscare movies")
plt.plot(years,non_oscar_revenue ,"<-y", label = "Non-Oscare movies")

plt.title("Oscar movies VS Non-Oscar moveis revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")

plt.legend()
```

```python
years = [2021,2022,2023,2024,2025]
revenue = [201,542,346,216,346]
non_oscar_revenue = [95,235,45,156,541]


plt.plot(years,revenue ,label = "Oscare movies")
plt.plot(years,non_oscar_revenue ,label = "Non-Oscare movies")

plt.title("Oscar movies VS Non-Oscar moveis revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")

print(plt.style.available)

# plt.style.use("dark_background")
plt.savefig("demo_fig.png")
```
