```python
s = pd.Series([1,2,3,4,5])
print(s)

#index
print(s[0])
print(s[4])
```

```python
s = pd.Series([24,25,98,48,65],index = ["Rahul","Geeta","Nikita","Sambhav","Kashi"])

print(s)
print(s["Rahul"])
print(s["Kashi"])
print(s.index)
```

```python
# Series are homogensous
s = pd.Series([1,2,3,4,"hello"])
print(s)
```

```python
# vertrizero operation
s1 = pd.Series([1,2,3,4])
s2 = pd.Series([1,2,3,4])
print(s1 + s2)

s1[0] = 100
chenged_s1 = s1.drop(0)
print(s1)
print(chenged_s1)
```
