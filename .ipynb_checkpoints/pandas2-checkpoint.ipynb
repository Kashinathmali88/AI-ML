```python
import pandas as pd
```

```python
# Cleaning Data
# Handling missing value

df = pd.read_csv("raw_data.csv")
df.isnull()
df.isna() # alies name for isnull()
df.isnull().sum() # chaning function
df.dropna() # drop row which has null value
df.dropna(axis = 1) # drop column
df.fillna(0)

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data
```
