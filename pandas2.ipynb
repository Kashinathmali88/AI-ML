```python
import pandas as pd
```

```python
# Cleaning Data
# Handling missing value

df = pd.read_csv("raw_data.csv")
df.isnull()
df.isna() # alies name for isnull()
df.isnull().sum() # chaning function
df.dropna() # drop row which has null value
df.dropna(axis = 1) # drop column
df.fillna(0)

cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)

df.ffill()
df.bfill()
```

```python
# handling Dupliction value
df["joining_date"].duplicated()
df[["department","joining_date"]].duplicated()
dfc = df.copy()
# df[["department","joining_date"]].drop_duplicates(inplace = True)
```

```python
# handling data types
df.dtypes
df["age"] = df["age"].fillna(0)
df["age"] = df["age"].astype("int64")
df.dtypes
df["joining_date"] = pd.to_datetime(df["joining_date"] , format="mixed", dayfirst=True, errors="coerce")
df["name"] = df["name"].astype("string")
df.dtypes
```

```python
# handling string
df["name"].str.upper()
df["name"].str.lower()
df["name"].str.capitalize()
df["name"].str.split(" ")
df["name"].str.contains("J" , case = False)
```
