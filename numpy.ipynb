```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
print(type(arr),arr.dtype,arr.size)

arr2 = np.array([1,2,3,4,5,"k"])
print(type(arr2),arr2.dtype)

arr3 = np.array([[1,2,3],[4,5,6]])
print(arr3.shape)
```

```python
arr4 = np.zeros((2,3),dtype = "int64")
print(arr4,arr4.shape)

arr4 = np.ones((2,3),dtype = "int64")
print(arr4,arr4.shape)

arr4 = np.full((2,3),200)
print(arr4,arr4.shape)

arr4 = np.eye((3),dtype = "int64")
print(arr4,arr4.shape)

arr4 = np.arange(1,10,2)
print(arr4,arr4.shape)

arr4 = np.linspace(1,100,4)
print(arr4,arr4.shape)
```

```python
#properties

arr = np.array([1,2,3,45,6,7,8])
print(arr.shape)
print(arr.size)
print(arr.dtype)
print(arr.ndim)

float_arr = arr.astype(np.float64)
print(float_arr,float_arr.dtype)
```

```python
#operation on array

a = np.array([[1,2,3],[4,5,6]])
print(a,a.shape)
#dimention change
reshape = a.reshape(3,2)
print(reshape,reshape.shape)

flatened = a.flatten()

print(flatened)
```

```python
#indexing
arr = np.array([1,2,3,4,5])
a = np.array([[1,2,3],[4,5,6]])

print(arr[0])
print(arr[1])
print(a[0,2])

#fancy indexing

ind = [1,2,3]
print(arr[ind])

#boolean indexing
print(arr[arr > 3])
print(arr[arr % 2 == 0])
```

```python
#slicing
print(arr[1:4])
print(arr[1:])
print(arr[:4])
print(arr[::2])
```
